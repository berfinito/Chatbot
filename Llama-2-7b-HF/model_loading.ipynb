```python
from transformers import AutoTokenizer, AutoModelForCausalLM

MODEL_ID = "meta-llama/Llama-2-7b-hf"

# 1️⃣ Tokenizer
tokenizer = AutoTokenizer.from_pretrained(
    MODEL_ID,
    use_auth_token=True    # tells HF to use your CLI‐logged token
)

# 2️⃣ Model
model = AutoModelForCausalLM.from_pretrained(
    MODEL_ID,
    use_auth_token=True,   # again, rely on the CLI login
    device_map="auto",     # auto‐place layers on GPU/CPU
    torch_dtype="auto"     # use fp16 if possible
)

print(f"✅ Loaded {MODEL_ID} (vocab_size={tokenizer.vocab_size})")
```
